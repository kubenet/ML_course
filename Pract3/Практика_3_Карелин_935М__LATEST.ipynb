##  Технические требования к генерации данных

```python
def generate_dataset(n_samples=500, n_features=2, noise_std=0.5,
                     outlier_fraction=0.05, outlier_multiplier=3,
                     random_seed=None):
    """
    Генерация синтетического датасета для линейной регрессии
    
    Parameters:
    -----------
    n_samples : int - количество наблюдений
    n_features : int - количество признаков
    noise_std : float - стандартное отклонение шума
    outlier_fraction : float - доля выбросов (0.0-0.2)
    outlier_multiplier : float - множитель для выбросов
    random_seed : int - seed для воспроизводимости
    """
    if random_seed is not None:
        np.random.seed(random_seed)
    
    # Генерация признаков
    X = np.random.randn(n_samples, n_features)
    
    # Истинные коэффициенты
    true_coef = np.random.uniform(1, 5, n_features)
    true_intercept = np.random.uniform(0, 10)
    
    # Генерация целевой переменной
    y = X @ true_coef + true_intercept + np.random.randn(n_samples) * noise_std
    
    # Добавление выбросов
    n_outliers = int(n_samples * outlier_fraction)
    outlier_idx = np.random.choice(n_samples, n_outliers, replace=False)
    y[outlier_idx] += np.random.choice([-1, 1], n_outliers) * \
                      np.random.uniform(3, 10, n_outliers) * outlier_multiplier
    
    return X, y, true_coef, true_intercept
```

````python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import (
    LinearRegression,
    HuberRegressor,
    SGDRegressor
)

from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

# Красивые графики ^_^
sns.set_style("whitegrid")
print("все готово брат")
````

генерация данных

````python
student_id = 6

features = ["reviews", "price"]

def generate_dataset(
    n_samples=500,
    n_features=2,
    noise_std=0.6,
    outlier_fraction=0.06,
    outlier_multiplier=3.5,
    random_seed=None
):
    """
    Генерация синтетического датасета для линейной регрессии
    """

    if random_seed is not None:
        np.random.seed(random_seed)

    # -----------------------------
    # 1. Генерация признаков
    # -----------------------------
    X = np.random.randn(n_samples, n_features)

    # -----------------------------
    # 2. Истинная линейная модель
    # -----------------------------
    true_w = np.random.uniform(-5, 5, n_features)
    true_b = np.random.uniform(-2, 2)

    # -----------------------------
    # 3. Шум
    # -----------------------------
    noise = np.random.normal(0, noise_std, n_samples)

    # -----------------------------
    # 4. Целевая переменная
    # -----------------------------
    y = X @ true_w + true_b + noise

    # -----------------------------
    # 5. Выбросы
    # -----------------------------
    n_outliers = int(outlier_fraction * n_samples)

    outlier_indices = np.random.choice(
        n_samples,
        n_outliers,
        replace=False
    )

    y[outlier_indices] *= outlier_multiplier

    return X, y, true_w, true_b




X, y, true_w, true_b = generate_dataset(
    n_samples=500,
    n_features=len(features),
    noise_std=0.6,
    outlier_fraction=0.06,
    outlier_multiplier=3.5,
    random_seed=student_id
)

print("Форма X:", X.shape)
print("Форма y:", y.shape)

print("\nИстинные коэффициенты:")
print(true_w)

print("\nИстинный bias:")
print(true_b)


df = pd.DataFrame(X, columns=features)
df["rating"] = y

print(df.head())
````

функция оценки и собсна оценка

````python
def evaluate_model(y_true, y_pred, model_name):

    mse = mean_squared_error(y_true, y_pred)

    rmse = np.sqrt(mse)

    mae = mean_absolute_error(y_true, y_pred)

    r2 = r2_score(y_true, y_pred)

    print("=" * 60)
    print(model_name)
    print("=" * 60)

    print(f"MSE  : {mse:.4f}")
    print(f"RMSE : {rmse:.4f}")
    print(f"MAE  : {mae:.4f}")
    print(f"R²   : {r2:.4f}")

    print("\n")

    return [model_name, mse, rmse, mae, r2]
    results = []

results.append(
    evaluate_model(
        y_test,
        linear_pred,
        "LinearRegression"
    )
)

results.append(
    evaluate_model(
        y_test,
        huber_pred,
        "HuberRegressor"
    )
)

results.append(
    evaluate_model(
        y_test,
        sgd_pred,
        "SGDRegressor"
    )
)
````

````python
models = {
    "LinearRegression": linear_pred,
    "HuberRegressor": huber_pred,
    "SGDRegressor": sgd_pred
}

results = []

for name, pred in models.items():

    residuals = y_test - pred

    results.append({
        "Model": name,
        "Mean Residual": np.mean(np.abs(residuals)),
        "Max Error": np.max(np.abs(residuals)),
        "Std Error": np.std(residuals)
    })

results_df = pd.DataFrame(results)

print(results_df)
plt.figure(figsize=(10, 5))

for name, pred in models.items():
    residuals = y_test - pred
    plt.hist(residuals, bins=40, alpha=0.4, label=name)

plt.axvline(0, color='red', linestyle='--')

plt.title("Distribution of residuals (robustness check)")
plt.xlabel("Error")
plt.ylabel("Frequency")

plt.legend()
plt.grid(alpha=0.3)

plt.show()
````

LinearRegression показывает более ровное распределение ошибок и меньшие экстремальные отклонения на данном наборе данных, однако это не означает универсально более высокую устойчивость, так как она чувствительна к выбросам и может сильно изменяться при изменении данных.

Наиболее устойчивой к выбросам оказалась модель HuberRegressor, так как использует комбинированную функцию потерь, уменьшающую влияние вбросов.

LinearRegression наиболее чувствительна к выбросам, так как минимизирует MSE.

SGDRegressor показывает промежуточное поведение и зависит от настройки параметров

влияние ЭПСИЛОН

````python
epsilons = [1.0, 1.35, 1.5, 2.0, 5.0]

results = []

for eps in epsilons:

    model = HuberRegressor(epsilon=eps, max_iter=1000)
    model.fit(X_train_scaled, y_train)

    pred = model.predict(X_test_scaled)

    mse = mean_squared_error(y_test, pred)
    mae = mean_absolute_error(y_test, pred)
    r2 = r2_score(y_test, pred)

    results.append([eps, mse, mae, r2])

results_df = pd.DataFrame(
    results,
    columns=["epsilon", "MSE", "MAE", "R2"]
)

print(results_df)
plt.figure(figsize=(10, 5))

plt.plot(results_df["epsilon"], results_df["MSE"], marker='o', label="MSE")
plt.plot(results_df["epsilon"], results_df["MAE"], marker='o', label="MAE")
plt.plot(results_df["epsilon"], results_df["R2"], marker='o', label="R2")

plt.xlabel("epsilon")
plt.title("влияние ЭПСИЛОН на производительность Huber")

plt.grid(alpha=0.3)
plt.legend()

plt.show()
````

Установлено, что при малых значениях epsilon модель становится более устойчивой к выбросам, но теряет точность на основном распределении данных.

При увеличении epsilon модель постепенно приближается к линейной регрессии и становится более чувствительной к выбросам.

Таким образом, epsilon управляет балансом между стабильностью и точностью модели.

ВЫВОДЫ:

LinearRegression эффективно работает на чистых данных, но чувствительна к выбросам.

HuberRegressor является наиболее устойчивой моделью при наличии шумов, так как сочетает квадратичную и линейную функции потерь.

SGDRegressor применяется для больших или потоковых данных, однако требует тщательной настройки гиперпараметров и может быть нестабильным.

Функция потерь MSE эффективно работает на данных без значительных выбросов и обеспечивает стабильную оптимизацию модели. За счёт квадратичного штрафа большие ошибки получают высокий вес, поэтому модель стремится минимизировать крупные отклонения. Однако высокая чувствительность к выбросам делает MSE менее устойчивой на шумных данных с аномальными значениями.

Huber loss сочетает преимущества MSE и MAE:

- для малых ошибок используется квадратичный штраф
- для больших ошибок — линейный

Это позволяет сохранять точность на основной массе данных и одновременно снижать влияние выбросов.

SGDRegressor использует стохастическую оптимизацию и может минимизировать разные функции потерь. При использовании MSE модель показывает результаты, близкие к обычной линейной регрессии, однако сильнее зависит от масштабирования данных и настройки гиперпараметров.
Стохастический подход делает модель более масштабируемой на больших данных, но повышает чувствительность к настройкам обучения.
